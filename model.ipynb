```python
monthly = df.groupby('Ticker').agg(
    Sector=('Sector','first'),
    First_Close=('Close Price','first'),
    Last_Close=('Close Price','last')
).reset_index()

monthly['Total_Return'] = monthly['Last_Close'] / monthly['First_Close'] - 1

```

```python
monthly
```

```python
monthly['Rank'] = monthly.groupby('Sector')['Total_Return'].rank(
    ascending=False, method='dense'
)

monthly['Label'] = (monthly['Rank'] == 1).astype(int)


```

```python
monthly.head(15)
```

```python
monthly
```

```python
monthly.head(20)

```

```python
monthly['Sector'].value_counts()
```

```python
df = df.merge(
    monthly[['Ticker', 'Total_Return', 'Rank', 'Label']],
    on='Ticker',
    how='left'
)
```

```python
df
```

```python
features = [
    'Open Price', 'Close Price', 'High Price', 'Low Price',
    'Volume Traded', 'Market Cap', 'PE Ratio', 'Dividend Yield',
    'EPS', '52 Week High', '52 Week Low'
]
```

```python
X = df[features]
y = df['Label']  
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Logistic Regression
log_model = LogisticRegression(max_iter=2000)
log_model.fit(X_train_scaled, y_train)
```

```python
# Decision Tree
tree_model = DecisionTreeClassifier(max_depth=5, random_state=42)
tree_model.fit(X_train, y_train)
```

```python
# SVM
svm_model = SVC(kernel='rbf', C=1)
svm_model.fit(X_train_scaled, y_train)
```

```python
y_pred_log = log_model.predict(X_test_scaled)
y_pred_tree = tree_model.predict(X_test)
y_pred_svm = svm_model.predict(X_test_scaled)
```

```python
print("\n📌 Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_log))
print(classification_report(y_test, y_pred_log))

print("\n📌 Decision Tree Accuracy:", accuracy_score(y_test, y_pred_tree))
print(classification_report(y_test, y_pred_tree))

print("\n📌 SVM Accuracy:", accuracy_score(y_test, y_pred_svm))
print(classification_report(y_test, y_pred_svm))
```
